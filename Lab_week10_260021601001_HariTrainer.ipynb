# Implementing Specification in Gemini

```python
import google.generativeai as genai
from google.api_core import exceptions

# System specification definition
system_instruction = """
System Specification: Barista Coffee Agent
1. Role & Persona:
You are an expert, welcoming Virtual Barista at "Wild Bean Leo Cafe". Your goal is to take customer orders, assist with menu choices, handle customizations, and ensure a seamless ordering experience.

2. Key Constraints & Formatting Rules:
- Tone: Warm, enthusiastic, polite, and conversational.
- Menu Options: Present options clearly when asked. Main categories include Espresso Drinks (Latte, Cappuccino, Americano), Cold Brew, and Teas (Matcha, Chai, Earl Grey).
- Customizations: Offer explicit choices for size (Small, Medium, Large), milk preference (Whole, Oat, Almond), sweetness level, and temperature (Iced or Hot).
- Allergy Warnings: Always prompt customers regarding potential allergens if they request nut-based milks or flavoured syrups.

3. Fallback & Edge Case Behaviours:
- Out-of-Menu Items: Politely explain that the item is unavailable and suggest the closest available menu alternative.
- Invalid/Ambiguous Inputs/ un related questions: Gently ask clarifying questions to resolve details before finalizing the order recap. politely ask them to enquire about the drink that they would like to have and say that the question they asked is invalid if it is not related.
"""

# Define generation parameters
generation_config = genai.types.GenerationConfig(
    temperature=0.2,
    max_output_tokens=4000
)

# Initialize model with system instruction
model = genai.GenerativeModel(
    'gemini-3.5-flash-lite',
    system_instruction=system_instruction,
    generation_config=generation_config
)

# Generate content with rate limit error handling
try:
    response = model.generate_content("Can I order a pumpkin spice frappuccino and a slice of avocado toast?")
    print(response.text)
except exceptions.ResourceExhausted as e:
    print("Rate limit exceeded. Please wait before trying again.")
except Exception as e:
    print(f"An error occurred: {e}")
```

# Implementing an Interactive Chat Agent in Python with Specification in Gemini


```python
import google.generativeai as genai
from google.api_core import exceptions
import requests

# System specification definition for Barista Coffee Agent
system_instruction = """
System Specification: Barista Coffee Agent
1. Role & Persona:
You are an expert, welcoming Virtual Barista at "Wild Bean Leo Cafe". Your goal is to take customer orders, assist with menu choices, handle customizations, and ensure a seamless ordering experience.

2. Key Constraints & Formatting Rules:
- Tone: Warm, enthusiastic, polite, and conversational.
- Menu Options: Present options clearly when asked. Main categories include Espresso Drinks (Latte, Cappuccino, Americano), Cold Brew, and Teas (Matcha, Chai, Earl Grey).
- Customizations: Offer explicit choices for size (Small, Medium, Large), milk preference (Whole, Oat, Almond), sweetness level, and temperature (Iced or Hot).
- Allergy Warnings: Always prompt customers regarding potential allergens if they request nut-based milks or flavoured syrups.

3. Fallback & Edge Case Behaviours:
- Out-of-Menu Items: Politely explain that the item is unavailable and suggest the closest available menu alternative.
- Invalid/Ambiguous Inputs/ un related questions: Gently ask clarifying questions to resolve details before finalizing the order recap. politely ask them to enquire about the drink that they would like to have and say that the question they asked is invalid if it is not related.
"""

# Configure model generation parameters
generation_config = genai.types.GenerationConfig(
    temperature=0.4,
    max_output_tokens=800
)

# Initialize Generative Model
model = genai.GenerativeModel(
    'gemini-3.5-flash-lite',
    system_instruction=system_instruction,
    generation_config=generation_config
)

# Start interactive chat session
chat = model.start_chat(history=[])
print("--- Barista Coffee Agent Initialized ---")
print("Type 'exit' or 'quit' to end the session.\n")

# Send initial welcome prompt to open dialogue
try:
    initial_response = chat.send_message("Hello! I'd like to order something to drink.")
    print(f"Barista: {initial_response.text}\n")
except exceptions.ResourceExhausted:
    print("Rate limit exceeded. Please wait a moment and try again.")
except Exception as e:
    print(f"An error occurred: {e}")

# Interactive Terminal Chat Loop
while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ['exit', 'quit']:
        print("Ending chat session. Goodbye!")
        break
    if not user_input.strip():
        continue
    try:
        response = chat.send_message(user_input)
        print(f"\nBarista: {response.text}\n")
    except exceptions.ResourceExhausted:
        print("\nRate limit exceeded. Please wait a bit before your next message.\n")
    except (requests.exceptions.Timeout, exceptions.DeadlineExceeded):
        print("\nRequest timed out. Please try sending your message again.\n")
    except Exception as e:
        print(f"\nAn error occurred: {e}\n")
```
